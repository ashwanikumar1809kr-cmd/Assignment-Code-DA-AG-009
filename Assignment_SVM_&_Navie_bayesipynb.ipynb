***Question 3:What is Pre-Pruning in Decision Trees?***

Answer:

Pre-pruning, also known as early stopping, is a technique used in decision trees to prevent overfitting by halting the tree growth early, before it becomes too complex. Instead of growing the tree fully and then trimming it (as in post-pruning), pre-pruning stops the tree from growing beyond a certain point based on specific conditions or constraints.

- Why is Pre-Pruning Important?

Decision trees, if left unregulated, can grow very deep and perfectly fit the training data. This often leads to overfitting, where the model performs well on training data but poorly on unseen data. Pre-pruning helps by simplifying the model and improving generalization.

- Common Pre-Pruning Techniques / Stopping Criteria

Maximum Depth

Limit how deep the tree can grow (e.g., stop after 5 levels).

Minimum Samples per Node

Only split a node if it contains at least a certain number of samples.

Minimum Information Gain (or Gini Reduction)

Only allow a split if it leads to a significant reduction in impurity.

Maximum Number of Nodes / Leaves

Restrict the total number of nodes or leaves in the tree.

Chi-Square Test (Statistical Significance)

Only split if the difference between classes is statistically significant.

- Advantages of Pre-Pruning

Reduces Overfitting: Simpler trees generalize better to new data.

Faster Training Time: Tree stops growing early, reducing computation.

Smaller Model Size: More efficient for deployment or storage.

- Disadvantages of Pre-Pruning

Risk of Underfitting: Stopping too early might prevent the tree from capturing important patterns.

May Miss Optimal Splits: Some useful splits might be skipped if thresholds are too strict.

- Example:

Suppose you are building a decision tree to classify emails as "spam" or "not spam".
If you set a minimum samples per node to 10, and at some point a node has only 8 samples, the algorithm will not split further, even if a split could improve accuracy. This is pre-pruning in action.

- Conclusion

Pre-pruning is a proactive strategy to control the complexity of a decision tree during its construction. By setting constraints like maximum depth or minimum samples, it helps in building more robust, faster, and generalizable models — though it must be used carefully to avoid underfitting.

***Question 4:Write a Python program to train a Decision Tree Classifier using Gini
Impurity as the criterion and print the feature importances (practical).
Hint: Use criterion='gini' in DecisionTreeClassifier and access .feature_importances_.
(Include your Python code and output in the code box below.)***

```python
# Import necessary libraries
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

# Load the Iris dataset
data = load_iris()
X = data.data
y = data.target
feature_names = data.feature_names

# Create and train the Decision Tree Classifier using Gini impurity
clf = DecisionTreeClassifier(criterion='gini', random_state=0)
clf.fit(X, y)

# Get feature importances
importances = clf.feature_importances_

# Print feature importances in a readable format
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

print("Feature Importances:")
print(importance_df)
```

***Question 5: What is a Support Vector Machine (SVM)?***

Answer:

A Support Vector Machine (SVM) is a powerful supervised machine learning algorithm used for classification and regression tasks. It works by finding the optimal boundary (hyperplane) that best separates the data into different classes.

- Core Idea Behind SVM

SVM tries to find a hyperplane in an N-dimensional space (where N is the number of features) that maximally separates the data points of different classes. The best hyperplane is the one that has the largest margin — the greatest distance between the hyperplane and the nearest data points from either class. These closest points are called support vectors, hence the name Support Vector Machine.

- Key Concepts
Term	Explanation
Hyperplane	A decision boundary that separates different classes. In 2D, it's a line; in 3D, it's a plane.
Margin	The distance between the hyperplane and the nearest data points (support vectors).
Support Vectors	The data points that are closest to the hyperplane and influence its position.
Kernel Trick	A method used to transform non-linearly separable data into a higher dimension where it becomes linearly separable.
-  Linear vs Non-linear SVM

Linear SVM: Used when data is linearly separable.

Non-linear SVM: Uses a kernel function (e.g., RBF, polynomial) to transform data into higher dimensions so it becomes linearly separable.

- Common Kernel Functions
Kernel Type	Description
Linear	For linearly separable data.
Polynomial	For curved boundaries.
RBF (Radial Basis Function)	Very popular; handles complex relationships well.
- Advantages of SVM

Works well with high-dimensional data

Effective when the number of features > number of samples

Robust to overfitting (especially in high dimensions)

Supports non-linear classification using kernels

- Disadvantages of SVM

Can be slow with large datasets

Not ideal for noisy data with overlapping classes

Choice of kernel and parameters can be tricky

- Example Use Case

SVMs are widely used in applications like:

Text classification (e.g., spam detection)

Image recognition

Medical diagnosis

Handwriting recognition

- Diagram (Optional for Your Assignment)

- Conclusion

Support Vector Machines are powerful tools for classification and regression. They work by finding the optimal hyperplane that separates data points of different classes with the maximum possible margin, making them effective for many real-world applications — especially when working with high-dimensional data.

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data points (two classes)
class1 = np.array([[2, 3], [3, 3], [2, 2]])
class2 = np.array([[7, 8], [8, 8], [7, 7]])

# Plot points
plt.scatter(class1[:, 0], class1[:, 1], color='blue', label='Class 1 (o)')
plt.scatter(class2[:, 0], class2[:, 1], color='red', label='Class 2 (x)')

# Draw hyperplane (decision boundary)
x = np.linspace(0, 10, 100)
# Example line: y = -x + 10
plt.plot(x, -x + 10, 'k-', label='Hyperplane')

# Draw margins
# Margin lines: y = -x + 12 and y = -x + 8 (parallel lines)
plt.plot(x, -x + 12, 'k--', label='Margin')
plt.plot(x, -x + 8, 'k--')

# Highlight support vectors (closest points)
support_vectors = np.array([[3, 3], [7, 7]])
plt.scatter(support_vectors[:, 0], support_vectors[:, 1], s=150, facecolors='none', edgecolors='green', linewidths=2, label='Support Vectors')

plt.title('SVM: Hyperplane, Margin, and Support Vectors')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

***Question 7: Write a Python program to train two SVM classifiers with Linear and RBF
kernels on the Wine dataset, then compare their accuracies.
Hint:Use SVC(kernel='linear') and SVC(kernel='rbf'), then compare accuracy scores after fitting
on the same dataset.
(Include your Python code and output in the code box below.)***

```python
# Import libraries
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# Load Wine dataset
wine = load_wine()
X = wine.data
y = wine.target

# Split dataset into training and testing sets (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Create SVM classifiers
svm_linear = SVC(kernel='linear', random_state=42)
svm_rbf = SVC(kernel='rbf', random_state=42)

# Train classifiers
svm_linear.fit(X_train, y_train)
svm_rbf.fit(X_train, y_train)

# Predict on test set
y_pred_linear = svm_linear.predict(X_test)
y_pred_rbf = svm_rbf.predict(X_test)

# Calculate accuracies
accuracy_linear = accuracy_score(y_test, y_pred_linear)
accuracy_rbf = accuracy_score(y_test, y_pred_rbf)

# Print results
print(f"Accuracy with Linear Kernel: {accuracy_linear:.4f}")
print(f"Accuracy with RBF Kernel: {accuracy_rbf:.4f}")

# Plotting accuracy comparison
kernels = ['Linear', 'RBF']
accuracies = [accuracy_linear, accuracy_rbf]

plt.bar(kernels, accuracies, color=['blue', 'green'])
plt.ylim([0, 1])
plt.ylabel('Accuracy')
plt.title('SVM Kernel Accuracy Comparison on Wine Dataset')
for i, v in enumerate(accuracies):
    plt.text(i, v + 0.02, f"{v:.2f}", ha='center', fontweight='bold')

plt.show()
```

***Question 9: Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve
Bayes, and Bernoulli Naïve Bayes***


Answer:

Naïve Bayes classifiers have different variants that are suited to different types of data and assumptions about feature distributions. The three common types are:

1. **Gaussian Naïve Bayes**

* Works with continuous features assumed to follow a normal (Gaussian) distribution.
* Calculates likelihoods using the probability density function of the Gaussian distribution for each feature.
* Suitable for continuous numerical data like height, weight, or any real-valued features.

2. **Multinomial Naïve Bayes**

* Works with discrete features representing counts or frequencies.
* Models features as multinomial distributions, e.g., counts of words in documents.
* Commonly used in text classification tasks such as document categorization and spam filtering.

3. **Bernoulli Naïve Bayes**

* Works with binary/Boolean features (presence or absence of a feature).
* Models features as binary variables (0 or 1), focusing on whether a feature occurs or not, regardless of count.
* Used for binary data such as whether a word occurs in a document or not.

**Summary:**

* Gaussian Naïve Bayes is best for continuous real-valued data.
* Multinomial Naïve Bayes is used for count data.
* Bernoulli Naïve Bayes is suited for binary/boolean data.

***Question 10: Breast Cancer Dataset
Write a Python program to train a Gaussian Naïve Bayes classifier on the Breast Cancer
dataset and evaluate accuracy.
Hint:Use GaussianNB() from sklearn.naive_bayes and the Breast Cancer dataset from
sklearn.datasets.
(Include your Python code and output in the code box below.)***

```python
# Import necessary libraries
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Load Breast Cancer dataset
data = load_breast_cancer()
X = data.data
y = data.target

# Split data into training and testing sets (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Create Gaussian Naive Bayes classifier
gnb = GaussianNB()

# Train the classifier
gnb.fit(X_train, y_train)

# Predict on test set
y_pred = gnb.predict(X_test)

# Evaluate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy of Gaussian Naive Bayes on Breast Cancer dataset: {accuracy:.4f}")

# Generate confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=gnb.classes_)

# Plot confusion matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=data.target_names)
disp.plot(cmap=plt.cm.Blues)
plt.title("Confusion Matrix: Gaussian Naive Bayes on Breast Cancer Dataset")
plt.show()
```
